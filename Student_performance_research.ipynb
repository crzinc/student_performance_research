```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Создаем целевой признак ПЕРЕД использованием
# Важно: убедись, что твой DataFrame называется df
df['is_zero'] = (df['G3'] == 0).astype(int)

# 2. Проверим корреляцию напрямую
# Выделяем только интересующие нас столбцы
alc_cols = ['Dalc', 'Walc', 'is_zero', 'G3', 'absences', 'failures']
correlation_matrix = df[alc_cols].corr()

print("Корреляция признаков с 'is_zero' (вероятностью получить 0):")
print(correlation_matrix['is_zero'].sort_values(ascending=False))

# 3. Визуализация: Влияние алкоголя на вероятность "вылета" (G3=0)
plt.figure(figsize=(12, 5))

# График для будней (Dalc)
plt.subplot(1, 2, 1)
sns.barplot(x='Dalc', y='is_zero', data=df, estimator=lambda x: sum(x)/len(x), palette='Reds')
plt.title('Вероятность G3=0 в зависимости\nот алкоголя в будни (Dalc)')
plt.ylabel('Доля студентов с нулями')

# График для выходных (Walc)
plt.subplot(1, 2, 2)
sns.barplot(x='Walc', y='is_zero', data=df, estimator=lambda x: sum(x)/len(x), palette='Blues')
plt.title('Вероятность G3=0 в зависимости\nот алкоголя в выходные (Walc)')
plt.ylabel('Доля студентов с нулями')

plt.tight_layout()
plt.show()
```

```python
# 1. Проверим, сколько реально человек в каждой группе
print("Количество студентов по уровням алкоголя в будни (Dalc):")
print(df['Dalc'].value_counts())

# 2. Посмотрим на средние оценки G1 и G2 у тех, кто пьет (уровни 3, 4, 5)
# Может они "талантливые прогульщики"?
heavy_drinkers_stats = df.groupby('Dalc')[['G1', 'G2', 'G3', 'absences']].mean()
print("\nСтатистика по группам Dalc:")
print(heavy_drinkers_stats)

# 3. Самый важный тест: Сравним тех, у кого G3=0 и G3>0 по пропускам
plt.figure(figsize=(8, 5))
sns.boxplot(x='is_zero', y='absences', hue='Dalc', data=df)
plt.title('Пропуски vs Нули в G3 (с разбивкой по алкоголю)')
plt.show()
```

```python
# 1. Считаем корреляцию для этой группы
health_alc_cols = ['Dalc', 'Walc', 'health', 'absences', 'G3']
corr_subset = df[health_alc_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_subset, annot=True, cmap='YlGnBu', fmt=".2f")
plt.title('Корреляция: Алкоголь, Здоровье, Прогулы и Оценки')
plt.show()

# 2. Визуализируем "Здоровье vs Прогулы" с разбивкой по алкоголю
# Используем Dalc (будни), так как он сильнее влияет на график
plt.figure(figsize=(12, 6))
sns.boxplot(x='health', y='absences', hue='Dalc', data=df)
plt.title('Влияние здоровья и алкоголя на количество прогулов')
plt.xlabel('Здоровье (1 - очень плохо, 5 - очень хорошо)')
plt.ylabel('Количество прогулов')
plt.legend(title='Алкоголь (будни)')
plt.show()

# 3. Линейный тренд: Как алкоголь и здоровье вместе "давят" на итоговый балл
sns.lmplot(x='health', y='G3', hue='Dalc', data=df[df['G3']>0], aspect=1.5, lowess=True)
plt.title('Зависимость G3 от здоровья при разном уровне алкоголя (без нулей)')
plt.show()
```
